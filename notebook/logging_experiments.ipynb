```python
import logging
import os
from datetime import datetime
```

```python
logs_dir = os.path.join(os.getcwd(), "logs_dir")
```

```python
logs_dir
```

```python
os.makedirs("logs_dir", exist_ok=True)
```

```python
LOG_FILE = f"{datetime.now().strftime('%Y-%m-%d_%H-%M-%S')}.log"
```

```python
#This is a log file(name of the log file have current date and time) which will be created in the logs_dir folder
LOG_FILE
```

#today
#with today's date log file will be created 

#tomorrow
#with tomorrow's date log file will be created

```python
print(logs_dir, type(logs_dir))
print(LOG_FILE, type(LOG_FILE))
```

```python
os.path.join(logs_dir, LOG_FILE)
```

```python
LOG_FILE_PATH = os.path.join(logs_dir, LOG_FILE)
```

```python
logging.basicConfig(
    filename=LOG_FILE_PATH,
    format="[%(asctime)s] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s",
    level=logging.INFO
)
```

```python
logging.getLogger("DocumentPortal")
```

```python
logger = logging.getLogger("DocumentPortal")
```

```python
logger.info("This is an info message")
```
