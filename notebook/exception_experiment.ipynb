```python
import sys
import traceback

class DocumentPortalException(Exception):
    """Custom exception class for Document Portal application."""
    def __init__(self, error_message, error_details: sys):
        _, _, exc_tb = error_details.exc_info()
        self.file_name = exc_tb.tb_frame.f_code.co_filename
        self.lineno = exc_tb.tb_lineno
        self.error_message = str(error_message)
        self.traceback_str = ''.join(traceback.format_exception(*error_details.exc_info()))
        
        super().__init__(self.error_message)  # ✅ Bug 2 fixed

    def __str__(self):
        return f"""
            Error in [{self.file_name}] at line: [{self.lineno}]
            Message: {self.error_message}
            Traceback: {self.traceback_str}
            """         # ✅ Bug 1 fixed — removed extra "


if __name__ == "__main__":
    from logger.custom_logger import CustomLogger  # ✅ Bug 3 fixed
    logger = CustomLogger().get_logger("exception_experiment")

    try:
        a = 1 / 0
        print(a)
    except Exception as e:
        app_exc = DocumentPortalException(e, sys)
        logger.error(str(app_exc))
        raise app_exc
```
